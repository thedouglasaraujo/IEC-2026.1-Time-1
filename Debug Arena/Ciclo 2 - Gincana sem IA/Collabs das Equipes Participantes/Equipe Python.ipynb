**Desafio 2 - Jogo da Forca**

**Jogo da Forca para Depuração** é uma adaptação educacional do tradicional jogo da forca, utilizada para desenvolver habilidades de identificação e correção de erros em programação. Nessa versão, em vez de adivinhar palavras, os participantes analisam trechos de código e tentam descobrir qual tipo de erro está presente, como SyntaxError, TypeError ou NameError.


```python
# Desafio 2

import random

codigos = [
    {
        "codigo": "for i in range(5)\nprint(i)",
        "erro": "IndentationError"
    },
    {
        "codigo": "x = input('Número: ')\nprint(x + 5)",
        "erro": "TypeError"
    }
]

vidas = 5
breakpoint()
desafio = random.choice(codigos)

print("=== DEBUGGING HANGMAN ===")

while vidas > 0:
    print("\nCódigo:")
    print(desafio["codigo"])

    resposta = input("\nQual o erro? ")

    if resposta.lower() == desafio["erro"].lower():
        print("✅ Correto!")
        break
    else:
        vidas -= 1
        print(f"❌ Errado! Vidas restantes: {vidas}")

if vidas == 0:
    print("☠ Game Over")
```

**Desafio 3 - Boss Battle**

**Boss Battle** é uma atividade gamificada inspirada nos confrontos finais dos jogos digitais, em que os participantes enfrentam um desafio de programação mais complexo que reúne diversos tipos de erros e problemas em um único cenário. Após resolver desafios menores ao longo da aprendizagem, os estudantes chegam ao "chefe final", representado por um programa com múltiplos bugs, falhas lógicas ou comportamentos inesperados. O objetivo é analisar o código, identificar os problemas e aplicar as correções necessárias para derrotar o desafio.

```python
# Desafio 3

boss_hp = 100

print("🐉 Infinite Loop Dragon apareceu!")

while boss_hp > 0:

    print(f"HP do Boss: {boss_hp}")

    resposta = input("Qual bug causa loop infinito? ")

    if "while true" in resposta.lower():
        dano = 25
        boss_hp -= dano
        print(f"⚔ Você causou {dano} de dano!")
    else:
        print("❌ Ataque falhou!")

print("🏆 Boss derrotado!")
```

**Desafio 4 - Quiz Battle**

**Quiz Battle** é uma atividade gamificada baseada em perguntas e respostas que desafia os participantes a testarem seus conhecimentos sobre programação, depuração e conceitos computacionais. Os estudantes competem individualmente ou em equipes, respondendo a questões relacionadas à identificação de erros, interpretação de códigos, boas práticas de desenvolvimento e resolução de problemas. A cada resposta correta, os participantes acumulam pontos, avançam em rankings ou desbloqueiam novas fases e desafios.

```python
# Desafio 4
perguntas = [
    {
        "codigo": "for i in range(5)",
        "resposta": "faltou dois pontos"
    },
    {
        "codigo": "print(x)",
        "resposta": "variável não definida"
    }
]

pontos = 0

for p in perguntas:
    breakpoint()
    print("\nCódigo:")
    print(p["codigo"])

    r = input("Qual o erro? ")

    if r.lower() == p["resposta"]:
        pontos += 10
        print("✅")
    else:
        print("❌")

print(f"\nPontuação final: {pontos}")
```

**Desafio 5 - RPG do Debugger**

**RPG do Debugger** é uma atividade gamificada que utiliza elementos dos jogos de interpretação de papéis (Role-Playing Game – RPG) para ensinar programação e depuração de código. Nessa abordagem, os estudantes assumem o papel de personagens especializados em solucionar problemas computacionais e enfrentam missões nas quais precisam investigar, identificar e corrigir erros em programas.

```python
# Desafio 5

xp = 0
nivel = 1

print("🧙 Bem-vindo ao RPG do Debugger")

while xp < 100:

    print(f"\nNível: {nivel}")
    print(f"XP: {xp}")

    resposta = input("Erro: NameError. O que significa? ")

    if "variável" in resposta.lower():
        xp += 20
        print("✅ XP ganho!")
    else:
        print("❌")

    if xp >= 50:
        nivel = 2

print("🏆 Você virou Mestre do Traceback!")
```

**IA Socrática para Debug**

**IA Socrática para Debug** é uma abordagem educacional que utiliza inteligência artificial como tutora virtual para apoiar o processo de depuração de código por meio do método socrático. Em vez de fornecer respostas prontas ou corrigir diretamente os erros, a IA conduz o estudante com perguntas, pistas e reflexões que estimulam a investigação e a descoberta da solução.

```python
import random

print("🧠 DEBUG ARENA - IA SOCRÁTICA")
print("=" * 40)

codigo = ""
numero = int(input("Digite um número: "))

if numero > 10:
    print("Maior")

print("\nCódigo com bug:")
print(codigo)

nivel_ajuda = 0

perguntas_socraticas = [
    "Qual é o tipo de dado retornado pelo input() em Python?",
    "O operador '>' consegue comparar tipos diferentes?",
    "O valor digitado pelo usuário é número ou texto?",
    "O que aconteceria se você convertesse o valor antes da comparação?"
]

dicas = [
    "💡 Dica 1: Observe o tipo da variável 'numero'.",
    "💡 Dica 2: input() retorna texto.",
    "💡 Dica 3: Talvez seja necessário usar int()."
]

while True:

    print("\nEscolha:")
    print("1 - Tentar resolver")
    print("2 - Pedir pergunta socrática")
    print("3 - Pedir dica")
    print("4 - Sair")

    opcao = input(">> ")

    if opcao == "1":

        resposta = input("\nComo você corrigiria o código?\n>> ")

        if "int" in resposta:
            print("n✅ Excelente!")
            print("Você identificou o problema de tipo de dados.")
            break
        else:
            print("\n🤔 Interessante hipótese.")
            print("Mas será que o problema está relacionado ao tipo da variável?")

    elif opcao == "2":

        pergunta = random.choice(perguntas_socraticas)
        print(f"\n🧠 IA Socrática pergunta:")
        print(pergunta)

    elif opcao == "3":

        if nivel_ajuda < len(dicas):
            print("\n" + dicas[nivel_ajuda])
            nivel_ajuda += 1
        else:
            print("\n⚠ Todas as dicas já foram utilizadas.")

    elif opcao == "4":
        print("\nEncerrando...")
        break

    else:
        print("\n❌ Opção inválida.")
```
