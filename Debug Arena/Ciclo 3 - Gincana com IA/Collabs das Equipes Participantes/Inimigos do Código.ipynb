A função abaixo calcula a **média das notas** de um aluno e retorna se ele foi `"aprovado"` (média >= 6) ou `"reprovado"`.

**Resultado esperado:**
```
→ {'media': 7.0, 'situacao': 'aprovado'}
```

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos antes de rodar.

````python
# Questão Exemplo
def calcular_media(notas):
    breakpoint()
    soma = 0
    for nota in notas:
        soma =+ nota

    media = soma / len(nota)

    if media >= 6:
        situacao = 'aprovado'
    else:
        situacao = 'reprovado'

    return {'media': media, 'situacao': situacao}

notas = [8, 6, 7, 9, 5]
print(calcular_media(notas))
````

## Questão 1

A função abaixo simula uma **caixa registradora**. Ela recebe uma lista de produtos
(cada um com `"nome"` e `"preço"`), um percentual de desconto e o valor pago.

Regras:
- O desconto só é aplicado se o total **bruto** for >= R$ 100,00.
- Retorna `{"total": ..., "troco": ...}`. Troco = valor pago − total final.

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ {'total': 99.0, 'troco': 1.0}
```


````python
## meu código:

def caixa_registradora(produtos, desconto_pct, valor_pago):
    total = 0

    for p in produtos:
        total += p["preco"]

    if total >= 100:
        total = total * (1 - desconto_pct / 100)
    else:
        pass

    troco = valor_pago - total

    return {
        "total": round(total, 2),
        "troco": round(troco, 2)
    }


produtos = [
    {"nome": "arroz",    "preco": 25.00},
    {"nome": "feijao",   "preco": 18.50},
    {"nome": "oleo",     "preco": 12.00},
    {"nome": "macarrao", "preco": 9.50},
    {"nome": "carne",    "preco": 45.00},
]

print(caixa_registradora(produtos, 10, 100.00))
````

## Questão 2

A função abaixo analisa a **frequência escolar** de uma turma.
- `"aprovado"` → frequência >= 75%
- `"reprovado"` → frequência < 75%

Se `"aulas_realizadas"` for 0, o aluno deve ser **ignorado**.

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ [
    {"nome": "Ana",    "frequencia": 75.0, "situacao": "aprovado"},
    {"nome": "Bruno",  "frequencia": 75.0, "situacao": "aprovado"},
    {"nome": "Carlos", "frequencia": 45.0, "situacao": "reprovado"},
]
```


***Código Correto***

````python

##Minha Resposta:
def analisar_frequencia(turma):
    resultado = []

    for aluno in turma:
        if aluno["aulas_realizadas"] == 0:
            pct = 0
        else:
            pct = (aluno["presencas"] / aluno["aulas_realizadas"]) * 100

        if pct >= 75:
            situacao = "aprovado"
        else:
            situacao = "reprovado"

        resultado.append({
            "nome": aluno["nome"],
            "frequencia": round(pct, 1),
            "situacao": situacao
        })

    return resultado


turma = [
    {"nome": "Ana",    "presencas": 15, "aulas_realizadas": 20},
    {"nome": "Bruno",  "presencas": 12, "aulas_realizadas": 16},
    {"nome": "Carlos", "presencas": 9,  "aulas_realizadas": 20},
    {"nome": "Diana",  "presencas": 0,  "aulas_realizadas": 0},
]

print(analisar_frequencia(turma))
````

## Questão 3

A função abaixo distribui **pontos de corrida** (menor tempo = melhor posição).

Tabela: 1º→25pts, 2º→18pts, 3º→15pts, 4º→12pts, 5º→10pts. Além do 5º: 0pts.

Empates: mesmo tempo = mesma posição. O próximo piloto após um grupo empatado
assume a posição seguinte ao grupo inteiro (ex.: dois em 1º → próximo é 3º).

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ [
    {"nome": "Alain",   "posicao": 1, "pontos": 25},
    {"nome": "Ayrton",  "posicao": 1, "pontos": 25},
    {"nome": "Michael", "posicao": 3, "pontos": 15},
    {"nome": "Lewis",   "posicao": 4, "pontos": 12},
    {"nome": "Max",     "posicao": 5, "pontos": 10},
    {"nome": "Nico",    "posicao": 6, "pontos": 0},
]
```


***Código Correto:***

````python
def pontuar_corrida(pilotos):
    tabela = {1: 25, 2: 18, 3: 15, 4: 12, 5: 10}
    ordenados = sorted(pilotos, key=lambda x: x["tempo"])
    resultado = []

    for i, piloto in enumerate(ordenados):
        if i == 0:
            posicao = 1
        elif piloto["tempo"] == ordenados[i - 1]["tempo"]:
            posicao = resultado[-1]["posicao"]
        else:
            posicao = i + 1

        pontos = tabela.get(posicao, 0)

        resultado.append({
            "nome": piloto["nome"],
            "posicao": posicao,
            "pontos": pontos
        })

    return resultado


pilotos = [
    {"nome": "Alain",   "tempo": 5400},
    {"nome": "Ayrton",  "tempo": 5400},
    {"nome": "Michael", "tempo": 5450},
    {"nome": "Lewis",   "tempo": 5480},
    {"nome": "Max",     "tempo": 5500},
    {"nome": "Nico",    "tempo": 5520},
]

for r in pontuar_corrida(pilotos):
    print(f"{r},")
````

## Questão 4

A função calcula o **saldo de cada pessoa** em uma viagem. Cada despesa tem
`"pagador"`, `"valor"` e `"participantes"` (incluindo o pagador). Custo dividido igualmente.

- Saldo positivo → deve receber. Saldo negativo → deve pagar.
- Despesas sem participantes devem ser **ignoradas**.

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ {'Ana': 185.0, 'Bruno': -100.0, 'Carlos': -85.0}
```


***Código correto:***

````python
def calcular_saldos(despesas):
    saldos = {}

    for despesa in despesas:
        valor = despesa["valor"]
        pagador = despesa["pagador"]
        participantes = despesa["participantes"]

        if len(participantes) == 0:
            continue

        parte = valor / len(participantes)

        if pagador not in saldos:
            saldos[pagador] = 0

        saldos[pagador] += valor

        for pessoa in participantes:
            if pessoa not in saldos:
                saldos[pessoa] = 0

            saldos[pessoa] -= parte

    return saldos


despesas = [
    {"descricao": "hotel",   "valor": 300.0, "pagador": "Ana",
     "participantes": ["Ana", "Bruno", "Carlos"]},
    {"descricao": "jantar",  "valor": 120.0, "pagador": "Bruno",
     "participantes": ["Ana", "Bruno", "Carlos"]},
    {"descricao": "passeio", "valor": 90.0,  "pagador": "Ana",
     "participantes": ["Ana", "Bruno"]},
    {"descricao": "gorjeta", "valor": 20.0,  "pagador": "Carlos",
     "participantes": []},
]

print(calcular_saldos(despesas))
````

## Questão 5

A função gerencia **emprestimos de biblioteca**.
Campos: `"livro"`, `"data_emprestimo"`, `"prazo_dias"`, `"renovacoes"`, `"data_devolucao"`.

Regras:
- Cada renovacao estende o prazo em `prazo_dias` a partir do prazo atual.
  (emprestado dia 1, prazo 7, 1 renovacao -> prazo ate dia 15)
- Multa: R$ 2,00 por dia **apos** o prazo final. No prazo (inclusive ultimo dia): multa = 0.

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ [
    {"livro": "Dom Casmurro", "prazo_final": 15, "multa": 2.0,  "status": "em atraso"},
    {"livro": "Memorias",     "prazo_final": 12, "multa": 0,    "status": "no prazo"},
    {"livro": "Iracema",      "prazo_final": 17, "multa": 2.0,  "status": "em atraso"},
]
```


***Código Correto***

````python
def processar_emprestimos(emprestimos):
    resultado = []

    for emp in emprestimos:
        prazo_final = (
            emp["data_emprestimo"]
            + emp["prazo_dias"] * (emp["renovacoes"] + 1)
        )

        dias_atraso = emp["data_devolucao"] - prazo_final

        if dias_atraso > 0:
            multa = dias_atraso * 2.0
            status = "em atraso"
        else:
            multa = 0.0
            status = "no prazo"

        resultado.append({
            "livro": emp["livro"],
            "prazo_final": prazo_final,
            "multa": multa,
            "status": status
        })

    return resultado


emprestimos = [
    {"livro": "Dom Casmurro", "data_emprestimo": 1,  "prazo_dias": 7,
     "renovacoes": 1, "data_devolucao": 16},
    {"livro": "Memorias",     "data_emprestimo": 5,  "prazo_dias": 7,
     "renovacoes": 0, "data_devolucao": 12},
    {"livro": "Iracema",      "data_emprestimo": 10, "prazo_dias": 7,
     "renovacoes": 0, "data_devolucao": 18},
]

for r in processar_emprestimos(emprestimos):
    print(r)
````

## Questão 6

A função simula um **mini-torneio**. Recebe jogadores (`"nome"`, `"pontos"` iniciais)
e rodadas — listas de partidas `("vencedor", "perdedor")`.

Regras: vitória = +1 ponto, derrota = +0 pontos.
Retorno: jogadores ordenados do **maior para o menor** número de pontos.

O código contém erros de **sintaxe**, **execução** e **lógica**. Encontre e corrija todos.

**Resultado esperado:**
```
→ [
    {"nome": "Ana",    "pontos": 2},
    {"nome": "Carlos", "pontos": 1},
    {"nome": "Diana",  "pontos": 1},
    {"nome": "Bruno",  "pontos": 0},
]
```


````python
def simular_torneio(jogadores, rodadas):
    pontos = {}

    for j in jogadores:
        pontos[j["nome"]] = j["pontos"]

    for rodada in rodadas:
        for vencedor, perdedor in rodada:
            if vencedor not in pontos:
                pontos[vencedor] = 0
            if perdedor not in pontos:
                pontos[perdedor] = 0

            pontos[vencedor] += 1

    resultado = [
        {"nome": nome, "pontos": pts}
        for nome, pts in pontos.items()
    ]

    resultado.sort(
        key=lambda x: (-x["pontos"], x["nome"])
    )

    return resultado


jogadores = [
    {"nome": "Ana",    "pontos": 0},
    {"nome": "Bruno",  "pontos": 0},
    {"nome": "Carlos", "pontos": 0},
    {"nome": "Diana",  "pontos": 0},
]

rodadas = [
    [("Ana", "Carlos"), ("Diana", "Bruno")],
    [("Ana", "Diana"),  ("Carlos", "Bruno")],
]

for j in simular_torneio(jogadores, rodadas):
    print(j)
````
